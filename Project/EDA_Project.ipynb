```python
numerical_cols = df.select_dtypes(
    include=[np.number]
).columns.tolist()

corr_matrix = df[numerical_cols].corr()
```

```python
print("\n--- Correlation Matrix ---")

print(corr_matrix)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(15, 15))

sns.heatmap(
    corr_matrix,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    vmin=-1,
    vmax=1,
    square=True,
    linewidths=0.5
)

plt.title(
    "Correlation Heatmap of Rainfall Features",
    fontsize=14,
    fontweight="bold"
)

plt.tight_layout()

plt.show()
```

```python
plt.figure(figsize=(15, 7))

sns.boxplot(
    data=df,
    x="SUBDIVISION",
    y="ANNUAL"
)

plt.title(
    "Annual Rainfall Distribution by Subdivision",
    fontsize=14,
    fontweight="bold"
)

plt.xlabel("Subdivision")
plt.ylabel("Annual Rainfall (mm)")

plt.xticks(rotation=90)

plt.tight_layout()

plt.show()
```

```python
import matplotlib.pyplot as plt

fig1 = plt.figure(figsize=(12, 6))

sns.heatmap(
    df.isnull(),
    cbar=False,
    cmap="viridis"
)

plt.title("Missing Values in Rainfall Dataset")

plt.xlabel("Features")
plt.ylabel("Records")

plt.show()
```

```python
features = [
    "JAN",
    "FEB",
    "MAR",
    "APR",
    "MAY",
    "JUN",
    "JUL",
    "AUG",
    "SEP",
    "OCT",
    "NOV",
    "DEC"
]

fig2 = plt.figure(figsize=(15, 8))

df[features].hist(
    bins=20,
    figsize=(15, 8)
)

plt.tight_layout()

plt.show()
```

```python
fig3 = plt.figure(figsize=(15, 8))

average_rainfall = df.groupby(
    "SUBDIVISION"
)["ANNUAL"].mean().sort_values(
    ascending=False
)

average_rainfall.plot(
    kind="bar"
)

plt.title(
    "Average Annual Rainfall by Subdivision"
)

plt.xlabel("Subdivision")

plt.ylabel(
    "Average Annual Rainfall (mm)"
)

plt.xticks(rotation=90)

plt.tight_layout()

plt.show()
```

```python
fig5 = plt.figure(figsize=(12, 10))

corr = df.select_dtypes(
    include=np.number
).corr()

sns.heatmap(
    corr,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title(
    "Correlation Heatmap of Rainfall Dataset"
)

plt.tight_layout()

plt.show()
```

```python
numerical_columns = [
    "JAN",
    "FEB",
    "MAR",
    "APR",
    "MAY",
    "JUN",
    "JUL",
    "AUG",
    "SEP",
    "OCT",
    "NOV",
    "DEC"
]

fig6 = plt.figure(figsize=(14, 8))

df[numerical_columns].boxplot()

plt.title(
    "Monthly Rainfall Distribution"
)

plt.xlabel("Month")

plt.ylabel(
    "Rainfall (mm)"
)

plt.xticks(rotation=45)

plt.tight_layout()

plt.show()
```

```python
yearly_rainfall = df.groupby(
    "YEAR"
)["ANNUAL"].mean()

plt.figure(figsize=(15, 6))

plt.plot(
    yearly_rainfall.index,
    yearly_rainfall.values
)

plt.title(
    "Annual Rainfall Trend in India"
)

plt.xlabel("Year")

plt.ylabel(
    "Average Annual Rainfall (mm)"
)

plt.grid(True)

plt.show()
```

```python
yearly_rainfall = df.groupby(
    "YEAR"
)["ANNUAL"].mean()

moving_average = yearly_rainfall.rolling(
    window=10
).mean()

plt.figure(figsize=(15, 6))

plt.plot(
    yearly_rainfall.index,
    yearly_rainfall.values,
    label="Annual Rainfall"
)

plt.plot(
    moving_average.index,
    moving_average.values,
    label="10-Year Moving Average"
)

plt.title(
    "Annual Rainfall Trend with 10-Year Moving Average"
)

plt.xlabel("Year")

plt.ylabel(
    "Rainfall (mm)"
)

plt.legend()

plt.grid(True)

plt.show()
```
